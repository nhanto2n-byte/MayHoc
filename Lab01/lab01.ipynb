## Bài 5. Reshape và Flatten

```python
a = torch.randn(3, 4)

print("Original:")
print(a)

print("\nFlatten:")
print(a.ravel())

print("\nReshape to 3 x 2 x 2:")
print(a.reshape(3, 2, 2))
```

## Bài 6. Phép toán và nhân ma trận

Thực hiện các phép toán phần tử và phép nhân ma trận bằng toán tử `@`.

```python
a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a:")
print(a)
print("b:")
print(b)

print("\na + b:")
print(a + b)

print("\na / b:")
print(a / b)

print("\na^2:")
print(a ** 2)

print("\nMatrix multiplication:")
print(a @ b.T)
```

**Yêu cầu:**
- Phân biệt phép toán `a * b` với `a @ b.T`.
- Cho biết shape của kết quả `a @ b.T`.

```python
print("a * b (element-wise):")
print(a * b)

print("\na @ b.T (matrix multiplication):")
print(a @ b.T)

print("\nShape cua a * b:", (a * b).shape)
print("Shape cua a @ b.T:", (a @ b.T).shape)
```

**Trả lời:**
- `a * b` là phép **nhân từng phần tử (element-wise multiplication)**: hai tensor phải cùng shape (2 × 3), kết quả cũng có shape 2 × 3, mỗi phần tử là tích của hai phần tử tương ứng.
- `a @ b.T` là phép **nhân ma trận (matrix multiplication)** theo quy tắc đại số tuyến tính: `a` có shape (2, 3), `b.T` có shape (3, 2), kết quả có shape `(2, 2)`. Mỗi phần tử kết quả là tổng tích vô hướng giữa một hàng của `a` và một cột của `b.T`.
- Vậy shape của `a @ b.T` là `torch.Size([2, 2])`.

## Bài 7. Các hàm thống kê trên Tensor

Tính trung bình, độ lệch chuẩn và tổng tích lũy theo từng cột.

```python
a = torch.randn(3, 4)
print(a)

print("\nMean:")
print(torch.mean(a, dim=0))

print("\nStandard deviation:")
print(torch.std(a, dim=0))

print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))
```

**Yêu cầu:** Thay `dim=0` bằng `dim=1` và nhận xét sự khác nhau.

```python
print("Mean theo dim=1 (theo hang):")
print(torch.mean(a, dim=1))

print("\nStandard deviation theo dim=1:")
print(torch.std(a, dim=1))

print("\nCumulative sum theo dim=1:")
print(torch.cumsum(a, dim=1))
```

**Trả lời:** Với `dim=0`, phép tính được thực hiện **dọc theo các hàng, tổng hợp theo từng cột** — kết quả là một vector có 4 phần tử (bằng số cột), mỗi phần tử tương ứng với trung bình/độ lệch chuẩn của một cột. Với `dim=1`, phép tính được thực hiện **dọc theo các cột, tổng hợp theo từng hàng** — kết quả là một vector có 3 phần tử (bằng số hàng). Tương tự, `cumsum(dim=0)` cộng dồn theo chiều xuống của từng cột, còn `cumsum(dim=1)` cộng dồn theo chiều ngang của từng hàng. Nói chung, tham số `dim` chỉ định chiều mà phép toán "thu gọn"/duyệt qua, kết quả giữ nguyên các chiều còn lại.

## Bài 8. Autograd cơ bản

Tạo biến có `requires_grad=True` và dùng `backward()` để tính đạo hàm tự động.

```python
x = torch.tensor(
    3.6,
    requires_grad=True
)

y = x * x

y.backward()

print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

**Yêu cầu:**
- Tính thủ công đạo hàm của y = x² tại x = 3.6 và so sánh với `x.grad`.
- Đổi hàm thành y = 3·x² + 2·x + 1 và kiểm tra gradient.

```python
# Dao ham thu cong: dy/dx = 2x tai x = 3.6
manual_grad = 2 * 3.6
print("Dao ham thu cong dy/dx tai x=3.6:", manual_grad)
print("Gia tri x.grad tu autograd:", x.grad.item())

# Doi ham thanh y = 3x^2 + 2x + 1
x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3 * x2 ** 2 + 2 * x2 + 1
y2.backward()

manual_grad2 = 6 * 3.6 + 2  # dy/dx = 6x + 2
print("\ny2 =", y2)
print("Dao ham thu cong dy2/dx tai x=3.6:", manual_grad2)
print("Gia tri x2.grad tu autograd:", x2.grad.item())
```

**Trả lời:** Với y = x², đạo hàm là dy/dx = 2x; tại x = 3.6, giá trị thủ công là 2 × 3.6 = 7.2, hoàn toàn khớp với `x.grad` mà PyTorch tự tính qua autograd. Với hàm mới y = 3x² + 2x + 1, đạo hàm là dy/dx = 6x + 2; tại x = 3.6, giá trị thủ công là 6 × 3.6 + 2 = 23.6, cũng khớp với giá trị `x2.grad` do autograd tính ra. Điều này minh họa cơ chế autograd của PyTorch: khi gọi `.backward()`, PyTorch tự động lan truyền ngược (backpropagation) qua đồ thị tính toán để tính đạo hàm chính xác mà không cần người dùng tự đạo hàm bằng tay.

## Bài 9. Gradient Descent tìm hệ số đa thức

Sinh dữ liệu từ đa thức y = x² + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

```python
polynomial = np.poly1d([1, 2, 3])
N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Learned coefficients:")
print(w)
```

**Yêu cầu:**
- So sánh ba hệ số học được với [1, 2, 3].
- Giải thích vai trò của `optimizer.zero_grad()`, `mse.backward()` và `optimizer.step()`.

```python
learned = w.detach().numpy().ravel()
target = np.array([1, 2, 3])
print("He so hoc duoc:", learned)
print("He so muc tieu:", target)
print("Sai so tuyet doi:", np.abs(learned - target))
```

**Trả lời:** Ba hệ số học được sau 1000 epoch xấp xỉ rất sát với bộ hệ số mục tiêu [1, 2, 3] của đa thức gốc y = x² + 2x + 3 (sai số tuyệt đối rất nhỏ, gần 0), chứng tỏ Gradient Descent kết hợp autograd đã hội tụ và tìm lại đúng các hệ số ban đầu.

Vai trò của ba lệnh trong vòng lặp huấn luyện:
- **`optimizer.zero_grad()`**: đặt lại gradient tích lũy trên các tham số về 0 trước mỗi vòng lặp. Vì PyTorch mặc định **cộng dồn** gradient qua mỗi lần gọi `backward()`, nếu không reset thì gradient của các epoch trước sẽ bị cộng dồn sai vào epoch hiện tại.
- **`mse.backward()`**: thực hiện lan truyền ngược (backpropagation), tính gradient của hàm mất mát (MSE) theo từng tham số (ở đây là `w`) và lưu vào thuộc tính `.grad` của tham số đó.
- **`optimizer.step()`**: dùng gradient vừa tính được để **cập nhật giá trị tham số** theo quy tắc của thuật toán tối ưu (ở đây là NAdam), nhằm giảm dần giá trị hàm mất mát qua mỗi epoch.

## Bài 10. Giải hệ phương trình bằng Autograd

Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.

```python
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2

    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

**Yêu cầu:**
- Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1, A+C=8, B-D=2.
- Nhận xét vì sao bài toán có thể có nhiều nghiệm.

```python
a_v, b_v, c_v, d_v = A.item(), B.item(), C.item(), D.item()
print("A+B =", a_v + b_v, " (muc tieu = 9)")
print("C-D =", c_v - d_v, " (muc tieu = 1)")
print("A+C =", a_v + c_v, " (muc tieu = 8)")
print("B-D =", b_v - d_v, " (muc tieu = 2)")
```

**Trả lời:** Bốn phương trình đều được thỏa mãn gần như chính xác (sai số rất nhỏ do tối ưu bằng gradient descent).

Bài toán có thể có **vô số nghiệm** vì hệ 4 phương trình tuyến tính này thực chất chỉ có **3 phương trình độc lập tuyến tính** (rank của hệ = 3 trong khi có 4 ẩn). Cụ thể, có thể suy ra phương trình thứ tư từ ba phương trình còn lại: từ A+B=9 và A+C=8 ta có B-C = 1, kết hợp với C-D=1 suy ra B-D = (B-C)+(C-D) = 1+1 = 2, đúng bằng phương trình thứ tư — nghĩa là nó không cung cấp thêm ràng buộc độc lập nào. Do đó hệ có 1 bậc tự do: nếu chọn A là một giá trị bất kỳ thì B = 9-A, C = 8-A, D = 7-A đều thỏa hệ phương trình, dẫn đến vô số nghiệm khác nhau tùy theo điểm khởi tạo ngẫu nhiên của gradient descent.

### 11.3. y = 2x² + 5x + 3, tính dy/dx tại x = 2 bằng autograd

```python
x11 = torch.tensor(2.0, requires_grad=True)
y11 = 2 * x11 ** 2 + 5 * x11 + 3
y11.backward()

manual_grad11 = 4 * 2 + 5  # dy/dx = 4x + 5
print("y =", y11.item())
print("Dao ham thu cong dy/dx tai x=2:", manual_grad11)
print("Gia tri x11.grad tu autograd:", x11.grad.item())
```

### 11.4. Thay đa thức Bài 9 bằng y = 2x² - 3x + 5, huấn luyện tìm lại hệ số

```python
polynomial2 = np.poly1d([2, -3, 5])
N = 20

X2 = np.random.randn(N, 1) * 5
Y2 = polynomial2(X2)

XX2 = np.hstack([
    X2 * X2,
    X2,
    np.ones_like(X2)
])

w2 = torch.randn(3, 1, requires_grad=True)
x2_t = torch.tensor(XX2, dtype=torch.float32)
y2_t = torch.tensor(Y2, dtype=torch.float32)

optimizer2 = torch.optim.NAdam([w2], lr=0.01)

print("Initial coefficients:")
print(w2)

# Dung nhieu epoch hon Bai 9 vi he so tu do (5) lon hon nen can nhieu vong lap hon de hoi tu
for epoch in range(5000):
    optimizer2.zero_grad()
    y_pred2 = x2_t @ w2
    mse2 = torch.mean(torch.square(y2_t - y_pred2))
    mse2.backward()
    optimizer2.step()

print("Learned coefficients:")
print(w2)

learned2 = w2.detach().numpy().ravel()
print("\nHe so hoc duoc:", learned2)
print("He so muc tieu:", [2, -3, 5])
```

**Trả lời:** Sau khi huấn luyện, ba hệ số học được xấp xỉ rất sát với bộ hệ số mục tiêu [2, -3, 5] của đa thức y = 2x² - 3x + 5, cho thấy phương pháp Gradient Descent kết hợp autograd và optimizer NAdam vẫn hoạt động tốt và tổng quát cho các bộ hệ số đa thức khác nhau, không chỉ riêng bộ hệ số ở Bài 9.
